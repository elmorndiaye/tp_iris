```python
import pandas as pd
import numpy as np
from sklearn.model_selection  import train_test_split # divise le dataset en donnee entrainement et test
from sklearn.naive_bayes import GaussianNB # algo du naive bayes
from sklearn.metrics import accuracy_score , confusion_matrix 
import pickle
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
```

```python
df = pd.read_csv("iris.csv")
df.head()
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
X = df.drop("Species" , axis = 1)
y = df["Species"]
X
```

```python
X_train, X_test ,y_train , y_test = train_test_split(X , y , test_size = 0.2)
```

## MIse en place de notre 1ere modele

```python
# chargement de l algo
model_NB = GaussianNB()
```

```python
# entraitenement
model_NB.fit(X_train , y_train )
```

```python
# Prediction
y_pred = model_NB.predict(X_test)
y_pred
```

```python
score =  cross_val_score(model_NB , X , y , cv = 5)
score.mean()
```
